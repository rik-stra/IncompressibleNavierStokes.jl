```julia
using DataFrames
using JLD2
using CairoMakie
using Printf
```

```julia
df_big = load(@__DIR__()*"/output/ks_data/ks_dists_range_1_4.jld2", "ks_table");
```

```julia
function f(x)
    if isnothing(x)
        return NaN
    else
        return x[1]
    end
end

df_big[!,"summed_ks"] = map(f, df_big[!,"ks_dist_ensemble"]);
```

```julia
function mi(x)
    if isnothing(x)
        return NaN
    else
        return minimum(stack(x)[1,:])
    end
end

function ma(x)
    if isnothing(x)
        return NaN
    else
        return maximum(stack(x)[1,:])
    end
end

df_big[!,"ks_best_replica"] = map(mi, df_big[!,"ks_dist_replicas"]);
df_big[!,"ks_worst_replica"] = map(ma, df_big[!,"ks_dist_replicas"]);
```
