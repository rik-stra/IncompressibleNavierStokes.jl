```julia
setups = []
times = []

open("time_results.txt") do f
    for line in eachline(f)
        sp = eachsplit(line, " ")
        push!(setups, collect(sp)[1])
        push!(times, parse(Float64, collect(sp)[end-2]))
    end
end
```

```julia
using DataFrames
using Statistics

df = DataFrame(setup = setups, time = times)
gdf = groupby(df, :setup)
means_df = combine(gdf, :time => mean)
```

```julia
function getdt(a)
    if a == "HIT_HF"
        return 0.00025
    elseif a == "Channel_HF"
        return 0.0005
    elseif a[1:10] == "Channel_LF"
        return 0.005
    elseif a[1:6] == "HIT_LF"
        return 0.0025
    end
end

function getSBU(s, setup)
    if setup in ["HIT_HF", "Channel_HF"]
        return s/3600 * 192
    else
        return s/3600 * 64
    end
end
```

```julia
means_df[!,"dt"] = getdt.(means_df[!,:setup])
means_df[!,"sec_per_timeunit"] = means_df[!, :time_mean]./(means_df[!, :dt]*100)
means_df[!,"SBU"] = getSBU.(means_df[!, :sec_per_timeunit], means_df[!, :setup])
means_df
```
